```python
import pandas as pd
df = pd.read_csv('/content/Sample_ Superstore.cvs.csv')
display(df.head())
```

```python
df.isnull().sum()
```

```python
df['Sales'] = df['Sales'].fillna(df['Sales'].median())
df['Region'] = df['Region'].fillna('Unknown')
```

```python
df = df.drop_duplicates()
```

```python
df['Category'] = df['Category'].str.title()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%m/%d/%Y', errors='coerce').dt.strftime('%Y-%m-%d')
```

```python
df.to_csv('cleaned_sales_data.csv', index=False)
```

```python
df = pd.read_csv('/content/cleaned_sales_data.csv')
```

```python
total_sales = df['Sales'].sum()
```

```python
avg_order_value = df['Sales'].mean()
```

```python
top_product = df.groupby('Category')['Sales'].sum().idxmax()
```

```python
print(f"Total Sales: ${total_sales:.2f}")
print(f"Average Order Value: ${avg_order_value:.2f}")
print(f"Top Product Category: {top_product}")
```

```python
sales_by_category = df.groupby('Category')['Sales'].sum().reset_index()
sales_by_region = df.groupby('Region')['Sales'].sum().reset_index()
```

```python
print("\n--- Sales by Category ---")
print(sales_by_category)
```

```python
print("\n--- Sales by Region ---")
print(sales_by_region)
```
